## Rescale the model from paper

```python
import sympy as sp
import numpy as np

sp.init_printing()
_tau, _xi, _kappa, _D, _Omega, _DOmega, _T = sp.symbols(r'\tau \xi \kappa D \Omega \dot{\Omega} T', real=True, positive=True)

kappa = sp.pi * 2 * _xi ** 2 / _tau

xi = sp.solve(kappa - _kappa, _xi)[0]
xi.subs({ _kappa: 9.96e-4, _tau: 1.0e-3 }).evalf()
```

```python
R = 230 * _xi
D = 2 * R
D.subs({ _xi: xi.subs({ _kappa: 9.96e-4, _tau: 1.0e-3 })}).evalf()
```

```python
Omega = 4e-3 * sp.pi/_tau
Omega.subs({_tau: 1.0e-3}).evalf()
```

```python
DOmega = 2.5e-8 * sp.pi/_tau**2
DOmega.subs({_tau: 1.0e-3}).evalf()
```

```python
vc = 0.2 * _xi / _tau
vc.subs({_xi: xi.subs({ _kappa: 9.96e-4, _tau: 1.0e-3 }), _tau: 1.0e-3}).evalf()
```
